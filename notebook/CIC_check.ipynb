```python
# 📘 CIC-IDS2018 - Phân tích file train/test theo Table 6

import pandas as pd
import os

# ====== 1. Cập nhật đường dẫn file CSV train & test ======
train_path = "../dataset/CSE-CIC-IDS2018/cic2018_training-set.csv"   # Đặt file trong cùng thư mục notebook
test_path  = "../dataset/CSE-CIC-IDS2018/cic2018_testing-set.csv"

# ====== 2. Load file ======
df_train = pd.read_csv(train_path)
df_test  = pd.read_csv(test_path)

# ====== 3. Tổng hợp nhãn gốc trong Label ======
train_counts = df_train["Label"].value_counts().to_frame(name="Training")
test_counts  = df_test["Label"].value_counts().to_frame(name="Testing")

# ====== 4. Gộp và tính tổng ======
summary = train_counts.join(test_counts, how="outer").fillna(0).astype(int)
summary["Total"] = summary["Training"] + summary["Testing"]
summary = summary.sort_values(by="Total", ascending=False)

# ====== 5. Hiển thị kết quả ======
print("\n\u2728 Bảng tổng hợp nhãn CIC-IDS2018 theo file train/test:\n")
print(summary)

```
