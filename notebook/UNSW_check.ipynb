```python
# 📘 UNSW-NB15 - Phân tích train/test dataset theo dạng bảng tổng hợp

import pandas as pd

# ====== 1. Cập nhật đường dẫn file train & test ======
train_path = "../dataset/unsw-nb15/unsw_nb15_training-set.csv"
test_path  = "../dataset/unsw-nb15/unsw_nb15_testing-set.csv"

# ====== 2. Đọc file CSV hoặc Excel ======
def load_file(path):
    if path.endswith(".csv"):
        return pd.read_csv(path, low_memory=False)
    elif path.endswith(".xlsx"):
        return pd.read_excel(path)
    else:
        raise ValueError("Chỉ hỗ trợ file CSV hoặc Excel")

# Đọc dữ liệu
train_df = load_file(train_path)
test_df  = load_file(test_path)

# ====== 3. Làm sạch cột ======
for df in [train_df, test_df]:
    df.columns = df.columns.astype(str).str.strip()
    df.drop(columns=[col for col in df.columns if col.startswith("Unnamed")], inplace=True, errors='ignore')

# ====== 4. Thống kê ======
if "attack_cat" in train_df.columns and "attack_cat" in test_df.columns:
    train_counts = train_df["attack_cat"].value_counts().to_frame(name="Training")
    test_counts  = test_df["attack_cat"].value_counts().to_frame(name="Testing")

    summary = train_counts.join(test_counts, how="outer").fillna(0).astype(int)
    summary["Total"] = summary["Training"] + summary["Testing"]
    summary = summary.sort_values(by="Total", ascending=False)

    print("\n📊 Bảng tổng hợp attack_cat theo Train/Test/Total:")
    print(summary)

elif "label" in train_df.columns and "label" in test_df.columns:
    train_counts = train_df["label"].value_counts().to_frame(name="Training")
    test_counts  = test_df["label"].value_counts().to_frame(name="Testing")

    summary = train_counts.join(test_counts, how="outer").fillna(0).astype(int)
    summary["Total"] = summary["Training"] + summary["Testing"]
    summary = summary.sort_values(by="Total", ascending=False)

    print("\n📊 Bảng tổng hợp label theo Train/Test/Total:")
    print(summary)

else:
    print("\n⚠️ Không tìm thấy cột 'label' hoặc 'attack_cat' để thống kê.")
```
